## 1. Importer les bibliothèques nécessaires
Importer `os` pour gérer les chemins et `nbformat` pour créer un notebook.

## 2. Créer un nouveau notebook dans l’espace de travail
Utiliser `nbformat` pour construire un notebook `.ipynb` dans le répertoire courant.

## 3. Ajouter une cellule de code
Ajouter une première cellule qui affiche `Hello, world!`.

## 4. Enregistrer et afficher le notebook créé
Écrire le notebook sur disque et afficher son chemin pour confirmer sa création.

```python
import os
import nbformat

notebook = nbformat.v4.new_notebook()
notebook.cells.append(
    nbformat.v4.new_code_cell("print('Hello, world!')")
)

output_path = os.path.join(os.getcwd(), "notebook_demo.ipynb")
nbformat.write(notebook, output_path)
print(f"Notebook créé : {output_path}")
```
